Importing the Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

Importing the Boston House Price Dataset

```python
house_price_dataset=sklearn.datasets.load_boston()
```

```python
print(house_price_dataset)
```

```python
#Loading the dataset to a Pandas dataframe
house_price_dataframe=pd.DataFrame(house_price_dataset.data, columns=house_price_dataset.feature_names)
```

```python
#Print the first 5 rows of the house_price_dataset
house_price_dataframe.head()
```

```python
#add the target(price) column to the dataset
house_price_dataframe['price']=house_price_dataset.target
```

```python
house_price_dataframe.head()
```

```python
#checking the number of rows and columns in the dataframe
house_price_dataframe.shape
```

```python
house_price_dataframe.isnull().sum()
```

```python
#the statistical measures for the dataset
house_price_dataframe.describe()
```

Understanding the corrrelation between various features in the dataset

```python
correlation=house_price_dataframe.corr()
```

```python
#constructing the heatmap to understand the correlation
plt.figure(figsize=(10,10))
sns.heatmap(correlation,cbar=True,square=True,fmt=".1f",annot=True, annot_kws={'size':8}, cmap='Blues')
```

Splitting the data and the target

```python
X=house_price_dataframe.drop(['price'],axis=1)
Y=house_price_dataframe['price']
```

```python
print(X)
print(Y)
```

Splitting the data into Training data and Test data

```python
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training

XGBoost Regressor

```python
#loading the model
model=XGBRegressor()
```

```python
#training the model with X_train, Y_train
model.fit(X_train,Y_train)
```

Evaluation

Prediction on training data

```python
#accuracy for prediction on training data
training_data_prediction=model.predict(X_train)
```

```python
print(training_data_prediction)
```

```python
#R square error
score_1=metrics.r2_score(Y_train,training_data_prediction)

#Mean absolute error
score_2=metrics.mean_absolute_error(Y_train,training_data_prediction)

print('R squared score: ',score_1)
print('Mean Absolute Error: ', score_2)
```

Visualizing the actual prices and the predicted values

```python
plt.scatter(Y_train,training_data_prediction)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual Price vs Predicted Price')
plt.show()
```

Prediction on Test data

```python
#accuracy for prediction on test data
test_data_prediction=model.predict(X_test)
```

```python
print(test_data_prediction)
```

```python
#R square error
score_3=metrics.r2_score(Y_test,test_data_prediction)

#Mean absolute error
score_4=metrics.mean_absolute_error(Y_test,test_data_prediction)

print('R squared score: ',score_3)
print('Mean Absolute Error: ', score_4)
```
